# Data Leaders 2.0: Machine-Learning Learning Roadmap

[Repository guide](../README.md) | [Next](./01_Exploratory_Data_Analysis.ipynb)

> Start here. The numbered notebooks form one complete beginner-friendly learning path.

This repository is not intended to be a collection of disconnected model
attempts. It is designed to teach **how to make an ML decision**.

Every phase compares a small number of meaningful methods under the same
validation setup and ends with a dataset-specific conclusion.

## The teaching pattern used in every phase

### Main method
The method receives a complete beginner-friendly explanation:

- what problem it solves;
- how it works conceptually;
- important parameters;
- advantages;
- disadvantages;
- leakage or misuse warnings;
- implementation;
- evaluation;
- conclusion on this dataset.

### Common alternatives
A few alternatives are implemented and compared fairly.

### Additional alternatives
Other important methods are named and explained briefly, but are not run
when they would only create repetitive experiments.

### Dataset example and justified decision
Every notebook includes a concrete example from this penalty-kick dataset and ends with:

- **technical evidence:** what the measured results show; and
- **theoretical reason:** why the method should or should not fit this problem.

## Phase map

| Phase | Main learning question |
|---|---|
| EDA and problem discovery | What did we find, what does it mean, and which notebook handles it? |
| Baselines and validation | What must an advanced method beat, and how are scores calculated? |
| Missing values and scaling | Which preprocessing is necessary for which model? |
| Categorical encoding | How can categories become model inputs without leakage? |
| Feature engineering | Can relationships be represented more clearly? |
| Feature selection | Which variables add useful predictive information? |
| Dimensionality reduction | Can encoded inputs be compressed without losing useful signal? |
| Outlier handling | Are unusual rows errors, noise, or useful hard cases? |
| Class imbalance | Does changing the training distribution improve the metric? |
| Model comparison | Which model family fits this tabular problem best? |
| Hyperparameter optimization | Which settings matter after choosing a model family? |
| Ensembling | Do different strong models make complementary errors? |
| Calibration and thresholds | Are probabilities reliable, and is 0.5 the right cutoff? |
| Interpretation and error analysis | Why does the model succeed or fail? |
| Final solution | What is the simplest strong end-to-end pipeline? |
| Approach comparison | How do local validation, offline test, and Kaggle scores differ? |

## Method-inclusion rule

A method belongs in the main comparison when it teaches a distinct idea.

A method belongs only in the alternatives section when:

- it is a small variation of a method already demonstrated;
- it is unlikely to suit this dataset;
- it adds substantial runtime without adding a new lesson;
- the comparison would require an unfair change to several pipeline stages.

This avoids a “method zoo” while keeping the repository useful as a
reference.

## Fair-comparison rule

During one experiment, keep everything fixed except the phase being tested.

Examples:

- When comparing encoders, keep the model and folds fixed.
- When comparing balancing methods, keep features, model, and folds fixed.
- When comparing model families, keep the feature set and folds fixed.
- When tuning a threshold, use out-of-fold probabilities rather than
  training predictions.

## Keep every experiment comparable

Every notebook loads the same competition files and records a small data
fingerprint before modeling. This prevents results from different dataset
versions from being mixed.

Before every experiment, record:

- train shape;
- test shape;
- target prevalence;
- feature columns;
- unique-ID checks.

Results from different fingerprints should be presented separately.
